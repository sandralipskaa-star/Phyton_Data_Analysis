```python
import ast
import matplotlib.pyplot as plt
import pandas as pd 
from datasets import load_dataset
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x:ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df['job_work_from_home'].value_counts().plot(kind='pie',startangle=90,autopct='%1.1f%%')
plt.title('Work from Home Status')
plt.ylabel('')
plt.show()
```

```python
df[['job_work_from_home','job_no_degree_mention','job_health_insurance']]
```

```python
fig, ax = plt.subplots(1,3)

dict_column={'job_work_from_home':'Work from Home','job_no_degree_mention':'No Degree Mention','job_health_insurance':'Health Insurance'}

for i,(column,title) in enumerate(dict_column.items()):
    ax[i].pie(df[column].value_counts(),labels=df[column].value_counts().index,autopct='%1.1f%%',startangle=90)
    ax[i].set_title(title)
```
