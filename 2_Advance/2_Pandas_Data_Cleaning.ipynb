```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
dataset = load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
```

```python
# task NAN value change for medium values, 
# remove duplicates#
#fillna, drop_duplicates
```

```python
df.loc[:10,'salary_year_avg':'salary_hour_avg']
```

```python
median_salary_year=df['salary_year_avg'].median()
```

```python
median_salary_hour=df['salary_hour_avg'].median()
```

```python
df_filled=df
```

```python
df_filled['salary_year_avg']=df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg']=df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
df_filled.loc[:,['salary_year_avg','salary_hour_avg']]
```

```python
df_unique=df_filled
```

```python
df_unique=df_unique.drop_duplicates()
print('Length before drop duplicates:',len(df_filled))
print('Length after drop duplicates:',len(df_unique))
print('Number of duplicates:',len(df_filled)-len(df_unique))
```

```python
df_unique_1=df_filled.drop_duplicates(subset=['job_title','company_name'])
print('Length before drop duplicates:',len(df_filled))
print('Length after drop duplicates:',len(df_unique_1))
print('Number of duplicates:',len(df_filled)-len(df_unique_1))
print('Number of duplicates:',len(df_unique)-len(df_unique_1))
```
