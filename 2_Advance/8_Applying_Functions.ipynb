```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
def projected_salary(row):
    if 'Senior' in row['job_title']:
        return row['salary_year_avg']*1.05
    else:
        return row['salary_year_avg']*1.03
        

df_salary['salary_year_inflated']=df_salary.apply(projected_salary,axis=1)
df_salary[['job_title','salary_year_avg','salary_year_inflated']]
```

```python
df_sal=df[pd.notna(df['salary_hour_avg'])].copy()

```

```python

df_sal['salary_hour_avg'].head()
```

```python
def pedicted(row):
    if 'Senior' in row['job_title']:
        return row['salary_hour_avg']*1.05
    else:
        return row['salary_hour_avg']*1.03

df_sal['salary_rate_inflated']=df_sal.apply(pedicted, axis=1)
df_sal['inflation%']=(df_sal['salary_rate_inflated']/df_sal['salary_hour_avg']-1)*100
df_sal[['job_title','salary_hour_avg','salary_rate_inflated','inflation%']]
```
